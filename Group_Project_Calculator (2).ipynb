## 1. Project Goal
The goal of this project is to build a multi-function calculator in Python. It allows a user to perform six math operations continuously untill they choose to exit.

## 2. Algorithm Description
* **Math Operations:** We created a `Calculator` class containing seperate methods for addition, substraction, multiplication, division, second power, and square root.
* **Continuous Loop:** A `while` loop displays the menu repeatedly after every calculation untill the user selects the exit option.
* **Conditional Logic:** We use `if` and `elif` statemnts to match the user´s menu choice to the correct math function and handle invalid inputs like dividing by zero.

```python
import math

class Calculator:
    # 1. Addition
    def add(self, a, b):
        return a + b

    # 2. Subtraction
    def subtract(self, a, b):
        return a - b

    # 3. Multiplication
    def multiply(self, a, b):
        return a * b

    # 4. Division
    def divide(self, a, b):
        if b == 0:
            return "Error: Cannot divide by zero!"
        return a / b

    # 5. Second Power (Square of a number)
    def second_power(self, a):
        return a ** 2

    # 6. Square Root
    def square_root(self, a):
        if a < 0:
            return "Error: Cannot calculate square root of a negative number!"
        return math.sqrt(a)
```

```python
# Create an object form the Calculator class
calc = Calculator()

# List of valid choices that need two numbers
two_number_ops = ["1", "2", "3", "4"]

while True:
    # Display the menu
    print("\n--- PYTHON CALCULATOR MENU ---")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Second Power")
    print("6. Square Root")
    print("7. Exit")

    choice = input("Choose an option (1-7): ")

    # Option 7: Exit the Loop
    if choice == "7":
        print("Exiting calculator. Goodbye!")
        break

    # Options 1 to 4: Operations requiring two numbers
    elif choice in two_number_ops:
        num1 = float(input("Enter First number: "))
        num2 = float(input("Enter second number: "))

        if choice == "1":
            print("Result:", calc.add(num1, num2))
        elif choice == "2": 
            print("Result:", calc.subtract(num1, num2))
        elif choice == "3":
            print("Result:", calc.multiply(num1, num2))
        elif choice == "4":
            print("Result:", calc.divide(num1, num2))

    # Option 5 and 6: Operations requiring one numbr
    elif choice == "5":
        num = float(input("Enter a number: "))
        print("Result:", calc.second_power(num))

    elif choice == "6":
        num = float(input("Enter a number: "))
        print("Result:", calc.square_root(num))

     # Handle wrong menu inputs
    else:
         print("Invalid choice! Please select a number between 1 and 7. ")
```
